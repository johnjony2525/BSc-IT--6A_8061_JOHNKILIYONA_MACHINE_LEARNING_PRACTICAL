```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler, OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_regression, f_classif
```

```python
df = pd.read_excel(r"/content/01 covid_toy.xlsx")
display(df.head())
```

```python
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

```

```python
print("Numberic Columns:")
print("numeric_columns")
```

```python
df["fever"] = pd.to_numeric(df["fever"], errors='coerce')
df["fever"] = df["fever"].fillna(df["fever"].mean())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
scaler = MinMaxScaler().set_output(transform="pandas")

scaled_data = scaler.fit_transform(df[numeric_columns])

scaled_data.head()
```

```python
# Define the target y (has_covid) and features X
y = df["has_covid"]
X = df.drop(columns=["has_covid"])

# Select only numeric features (age, fever)
X_numeric = X.select_dtypes(include=np.number)

# Perform feature selection using f_classif (since target is categorical)
selector = SelectKBest(
    score_func=f_classif,
    k=2
).set_output(transform="pandas")

X_selected = selector.fit_transform(X_numeric, y)

print(X_selected.head())
```
