```python
import numpy as np
import pandas as pd
```

```python

from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder

from sklearn.feature_selection import SelectKBest, f_regression, f_classif
```

```python
df = pd.read_csv("/content/auto-mpg (1).csv")

display(df.head())
```

```python
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
```

```python
print("numeric_columns")
print(numeric_columns)
```

```python
# Correcting the previous copy-paste error to identify actual categorical (object) columns
categorical_columns = df.select_dtypes(include=['object']).columns
```

```python
print("categorical:")
print(categorical_columns)
```

```python
print(df.isnull().sum())
```

```python
df["horsepower"] = df["horsepower"].replace('?', np.nan)
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
print("Horsepower column processed successfully. New data types:")
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
scaler = MinMaxScaler().set_output(transform="pandas")
scaled_data = scaler.fit_transform(df[numeric_columns])
```

```python
scaled_data.head()
```

```python
x = df.drop("mpg", axis=1)
y = df["mpg"]
```

```python
X_numeric = x.select_dtypes(include=np.number)

selector = SelectKBest(
    score_func=f_regression,
    k=5
).set_output(transform="pandas")

X_select = selector.fit_transform(X_numeric, y)

print(X_select.head())
```
